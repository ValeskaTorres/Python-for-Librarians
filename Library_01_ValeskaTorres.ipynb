```python
#A small exercise set in a library
print("CIRCULATION DESK")
username = input("User name: ")
book = input("Book title: ")
overdue_days = int(input("Days overdue: "))
daily_fee = float(input("Daily fine amount: $ "))

print("\nSUMMARY")
print(f"User: {username}")
print(f"Book: {book}")

if overdue_days > 0:
    fine = overdue_days * daily_fee
    print(f"Status: {overdue_days} day(s) overdue.")
    print(f"Estimated fine: $ {fine:.2f}")
else:
    print("Status: loan is on time.")
    print("Estimated fine: $ 0.00")
```
